Find the similarity of the word by using Word2Vec Cbow & Skip gram method

## Data Collection

Downloaded poniyin selvan story book as pdf 

```python
pdf_path = r"R:\Learning\Core_NLP\02_Word2Vec\Poniyan_Selvan_Story.pdf"
```

## Data understanding

### Import necessary libraries

```python
from pypdf import PdfReader
import re
from gensim.models import Word2Vec
from nltk.corpus import stopwords
import nltk
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
reader = PdfReader(pdf_path)

raw_text = ""

for page in reader.pages:
    text = page.extract_text()

    if text:
        raw_text += text + " "

print("Total Characters:", len(raw_text))
print("\nSample Text:\n")
print(raw_text[:100])
```

## Data pre-processing

### Tokenizaton

```python
# Sentence tokenization
sentences_raw = re.split(r'[.!?]', raw_text)
sentences = []

stop_words = set(stopwords.words('english'))

for sent in sentences_raw:

    sent = sent.lower()

    sent = re.sub(r'[^a-zA-Z\s]', ' ', sent)

    sent = re.sub(r'\s+', ' ', sent).strip()

    # Word tokenziation
    words = sent.split()

    # Remove stopwords
    words = [word for word in words if word not in stop_words]
    
    if len(words) > 1:
        sentences.append(words)
```

 Note : Vocabulary creation handled internall by word2Vec . Word tokenization is enough in pre-processing

## Model Building

```python
# Continuos Bag of words method (CBow)

model_cbow = Word2Vec(sentences=sentences,
                 vector_size=100,
                 window=5,
                 min_count=1,
                 workers=4,
                 epochs=20,
                 sg=0
)
print("\nModel Trained Successfully")
```

```python
# Continuos Bag of words method (Skip Gram)

model_skip_gram = Word2Vec(sentences=sentences,
                 vector_size=100,
                 window=5,
                 min_count=1,
                 workers=4,
                 epochs=20,
                 sg=1
)
print("\nModel Trained Successfully")
```

## Model Testing

### CBOW Test with single word

```python
word = "aditya"

if word in model_cbow.wv:

    print(f"\nWords similar to '{word}'")

    similar_words = model_cbow.wv.most_similar(word, topn=10)

    for w, score in similar_words:
        print(w, round(score, 4))

else:
    print(f"{word} not found in vocabulary.")
```

### CBOW Test with Sentence

```python
sample_sentence = "the king lived in a palace"

test_words = sample_sentence.lower().split()

print("\nTesting Sentence:")
print(sample_sentence)

for word in test_words:

    if word in model_cbow.wv:

        print(f"\nSimilar words for: {word}")

        for w, score in model_cbow.wv.most_similar(word, topn=5):
            print(f"{w:15} {score:.4f}")
```

### SKIP GRAM Test with single word

```python
word = "aditya"

if word in model_skip_gram.wv:

    print(f"\nWords similar to '{word}'")

    similar_words = model_skip_gram.wv.most_similar(word, topn=10)

    for w, score in similar_words:
        print(w, round(score, 4))

else:
    print(f"{word} not found in vocabulary.")
```

### SKIP GRAM Test with Sentence

```python
sample_sentence = "the king lived in a palace"

test_words = sample_sentence.lower().split()

print("\nTesting Sentence:")
print(sample_sentence)

for word in test_words:

    if word in model_skip_gram.wv:

        print(f"\nSimilar words for: {word}")

        for w, score in model_skip_gram.wv.most_similar(word, topn=5):
            print(f"{w:15} {score:.4f}")
```
